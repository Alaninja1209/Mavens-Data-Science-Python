## Foundations: Definitions and Syntax

This notebook contains some definition concepts that will help you in your day-to-day Data Scientist and a good recap on syntax and naming conventions

```python
'''
A programming language consists of words and symbols we use to write instructions for computers
to follow, while libraries are reusable collection code
'''

# The = is use specifically to assign values to a variable
age = 14
print(age)

# The == is for comparison, evaluates if both variables have the same value
print(age == 1000)
print(10 * 3 == 30)
```

```python
# Python allows us to perform actions based on conditional logic
if age >= 18:
    print("Adult")
else:
    print("Minor")
```

```python
# Another common tasks is looping, which performs the same action to each element of something
for number in [1, 2, 3, 4, 5]:
    print(number, end = " ")
```

```python
my_list = [5, 10, 15, 20, 25]
for n in my_list:
    print(n / 5, end = " ")
```

```python
'''
If we want to perform the same task multiple times. For best practices, it's recommended to implement
a function which is a chunk of code that cna be reused to perform the same tasks
'''

# Implementing age function
def is_adult(age):
    if age >= 18:
        print("Adult")
    else:
        print("Minor")

is_adult(20)
```

```python
'''
Python has its own library of built in functions to perform common tasks, such as the sorted
function who orders the list in ascending or descending order
'''

# Sorted function to obtain an ordered list
new_list = [20, 5, 50, 45, 30]

# Reverse parameter allows you to order list in ascending or descending
sorted(new_list, reverse = False)
```

## Object Oriented Programming Concepts

Definition: A programming system that is based around objects, which can contain both data and code that manipulates data
- Object: Instance of a class, fundamental building block of Python (Strings and Lists are objects)

Have data and methods define around objects to make the code more reusable, organize and accesible.
- Class: Object's data type that bundles data and functionality together. Useful to build tools that can be packed in the object itself

```python
# This variable become an instance of the string class
magic = 'HOCUS POCUS'
print(type(magic))
```

```python
# Turning lowercases to uppercases and vice-versa
magic = magic.swapcase()
magic
```

```python
# Replacing characters in string
magic = magic.replace('cus', 'key')
magic
```

```python
# Splitting the strings into a list of two strings
magic = magic.split()
magic
```

The actions done above are known as methods, a function that belongs to a class and typically perfoms an action or operation.

We use dot notation to access the methods and attributes that belong to an instance of a class. Some core classes of Python are:
- Integers
- Strings
- Floats
- Lists

We have attributes too, which are associated values with an object or class which is referenced by name using dot notation. Access characteristics of a class, but you can't do anything to change it.

**Attributes are characteristics and methods are actions or operations**

### Python Syntax and Variables

A data type is an attribute that describes a piece of data based on its value, its programming language or the operations it can perform.

Naming variables is important to let know others what its stored in the variable.
- Assignment: Process of storing a value in a variable
- Expression: A combination of numbers, symbols, or other variables that produce a result when evaluated

```python
age_list = [34, 25, 23, 19, 32]

# Get maximum value from our age list
max_age = max(age_list)
max_age
```

```python
# If we want our max age to become a string value, we can turn it
max_age = str(max_age)
max_age

# Getting the minimum age and getting the difference
int(max_age) - min(age_list)
```

We may encounter with **Naming conventions** which are consistent guidelines that describe the content, creation date, and version of a file in its name and **Naming restrictions** rules that built into syntax of the language itself that must be followed.

- Keyword: Special word that is reserved for a specific purpose and that can only can be used for that purpose (Ex. else or if)

```python
'''
When performing mathematical operations, Python follows the order of operations (PEMDAS), use 
parentheses to define priority, as operations inside them are performed first.
'''

print(2 * 3 + 4)
print((3 + 4) * 3)
```

```python
'''
Text written between quotes are known as strings, known as a sequence of characters and punctuation 
that contains textual information. They're an immutable type of data, which means values they 
can never be altered or updated.

Integers are data types used to represent whole numbers without fractions and floats represent 
numbers that contain decimals
'''

# Use the type function to have the computer tell you the data type
type('A')
```

Consider the **Implicit Conversion** where Python automatically converts data type to another 
without user involment

```python
# Mathematical operations, integers are turned to floats
print(8 + 1.5)
```

```python
'''
To avoid this, do a explicit conversion where you turn the data type of an object to
the required type

This is called typecasting as you're changing the data type
'''

# The output of this will be save as a string
print('2 + 2 = ' + str(2 + 2))
```
