### Foundations: Loops and Strings

This notebook contains the explanation and iterations that we can perform as loops and the basic operations that we can apply on strings

```python
'''
Loops are a block of code used to carry out iterations which are repeated execution of statements,
where one iteration is the single execution of a block of code

Iterable is an object that can be looped
'''

# While loop executes code based on the value of a condition evaluated to true or false
x = 0
while x < 5:
    print(x, end = " ")
    x += 1

print("x =", str(x))
```

```python
# We can end a loop even if the evaluation is true, use break
x = 1
i = 0

while x < 100:
    if i == 5:
        break
    print(i, x, end = " ")
    x = x * 2
    i += 1
```

```python
# We can skip the execution of the rest of the code with the continue statement
i = 0

while i < 10:
    if i % 3 != 0:
        print(i, end = " ")
        i += 1
        continue

    i += 1
```

```python
'''
For loops iterate over a sequence of values, the range function returns a sequence of numbers
starting from 0 and stops when it reach the given number.

The list of numbers generated will be one less than the given value and start on 0
'''
for x in range(5):
    print(x, end = " ")
```

```python
# We can use a for loop to read the content of a text file
with open(r'C:\Users\alana\Documents\My_stuff\Data_Scientist_Training\Mavens-Data-Science-Python\Foundations\foundation_files\Limit_X_Survivor.txt') as f:
    for line in f:
        print(line)

print("Done!")
```

```python
'''
When using the range method on a for loop, we can send a start or end loop through parameters
'''

# First parameter is start point, and the other end point
product = 1
for n in range(1, 10):
    product = product * n

print(product)
```

```python
def to_celsius(x):
    return (x - 32) * (5 / 9)

# The last parameter represents the steps to take in the loop
for x in range(0, 101, 10):
    print(x, to_celsius(x))
```

Use **while loops** when you want to repeat an action until a boolean condition changes, and use a **for loop** when there's a sequence of elements that you want to iterate over

```python
'''
Nested loops helps us to extract information from nested structures (list of lists), becareful of
adding a lot it could cause our code to look confusing
'''

hogwarts_houses = [['Harry', 'Ron'], ['Malfoy', 'Goyle']]

for house in hogwarts_houses:
    for student in house:
        print(student)
```

### Strings and some theory

Sequence of charcaters and punctuation that contains textual information, it's an unmmutable data type which means it can't be modified. Some actions that we can perform are:
- Concatenate: Join together. make a longer string 

Indexing is a way to refer the individual items within an iterable by their relative position

```python
# Performing concatenation
'Hello ' + 'world'
```

```python
# Multiplying strings
'Danger!' * 3
```

```python
# Quotation marks indicates end and start of a string, you can use backslash
quote = '"I am Lord Voldemort"'
print(quote)

profecy = "\"The one with the power to defeat Voldemort is coming\""
print(profecy)

# Backslash-n to indicate a new line
killing_curse = 'Avada Kedavra, \nVoldemort'
print(killing_curse)
```

```python
# We can iterate over a string too
phrase = 'Pikachu'

for char in phrase:
    print(char + 'at')
```

```python
# Index method outputs it's index number
pets = 'cats and dogs'
pets.index('n')
```

```python
# Use index number to find specific character in a string
name = 'Salazar Slytherin'
name[0]
```

```python
# We can access last element of a string through negative indices
name[-1]
```

```python
'''
String slicing is a portion of a string, better known as substring, it can have more than
one character
'''

# Obtaining subtring from 1st character to fourth character
name[1:5]
```

```python
'''
You can put as end point a huge point that surpasses the string length, it'll 
return it till last index
'''

# Here we start from the first index and going till the end
name[:7]
```

```python
# Use keyword in to check if substring is in the string
'Slytherin' in name
```

```python
'''
Formay method formats and inserts specific subtrings in designated places
'''

magician = 'Hermoine'
death_eater = 'Bellatrix'
print('{} is facing {} in a final duel'.format(magician, death_eater))
```

```python
player = 'Giancarlo'
number = 5
print('This is the {num} try of {player} to defeat Master Hand'.format(num = number, player = player))
```

```python
player = 'Alan'
number = 1
print('This is the {1} try of {0} to defeat Master Hand'.format(player, number))
```

```python
# Use the .2f format to indicate number of decimals you want to put in the string

price = 5.00
after_tax = 7.00
print('Base price ${:.2f} USD, after tax is ${:.2f} UDS'.format(price, after_tax))
```

```python
'''
Use the :>3 to align text to the right by three spaces, you can indicate the number of
spaces by changing the number
'''

for x in range(0, 101, 10):
    print("{:>3} F | {:>6.2f} C".format(x, to_celsius(x)))
```

### Formatting options

To use the formatting options put your string in braces and consider the following points:
- The variable that is being formatting
- The colon separates whta's being formatted from the syntax used to format it 
- .number indictes the precision
- The letter indicates presentation type

`{float:.2f}`

**f** represents fixed-point notation, formats the number as decimal number with the decimals specified. **%** multiplies the number by 100 and displays it following the % sign and finally, **e** formats the number in scientific notation separating the coefficient from the exponent
