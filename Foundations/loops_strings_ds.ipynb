### Foundations: Loops and Strings

This notebook contains the explanation and iterations that we can perform as loops and the basic operations that we can apply on strings

```python
'''
Loops are a block of code used to carry out iterations which are repeated execution of statements,
where one iteration is the single execution of a block of code

Iterable is an object that can be looped
'''

# While loop executes code based on the value of a condition evaluated to true or false
x = 0
while x < 5:
    print(x, end = " ")
    x += 1

print("x =", str(x))
```

```python
# We can end a loop even if the evaluation is true, use break
x = 1
i = 0

while x < 100:
    if i == 5:
        break
    print(i, x, end = " ")
    x = x * 2
    i += 1
```

```python
# We can skip the execution of the rest of the code with the continue statement
i = 0

while i < 10:
    if i % 3 != 0:
        print(i, end = " ")
        i += 1
        continue

    i += 1
```

```python
'''
For loops iterate over a sequence of values, the range function returns a sequence of numbers
starting from 0 and stops when it reach the given number.

The list of numbers generated will be one less than the given value and start on 0
'''
for x in range(5):
    print(x, end = " ")
```

```python
# We can use a for loop to read the content of a text file
with open(r'C:\Users\alana\Documents\My_stuff\Data_Scientist_Training\Mavens-Data-Science-Python\Foundations\foundation_files\Limit_X_Survivor.txt') as f:
    for line in f:
        print(line)

print("Done!")
```

```python
'''
When using the range method on a for loop, we can send a start or end loop through parameters
'''

# First parameter is start point, and the other end point
product = 1
for n in range(1, 10):
    product = product * n

print(product)
```

```python
def to_celsius(x):
    return (x - 32) * (5 / 9)

# The last parameter represents the steps to take in the loop
for x in range(0, 101, 10):
    print(x, to_celsius(x))
```

Use **while loops** when you want to repeat an action until a boolean condition changes, and use a **for loop** when there's a sequence of elements that you want to iterate over

```python
'''
Nested loops helps us to extract information from nested structures (list of lists), becareful of
adding a lot it could cause our code to look confusing
'''

hogwarts_houses = [['Harry', 'Ron'], ['Malfoy', 'Goyle']]

for house in hogwarts_houses:
    for student in house:
        print(student)
```
