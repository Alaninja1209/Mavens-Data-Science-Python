### Writing clean code

For best practices, it is recommend to create functions common repeated tasks to use it many times without rewriting it. This will support on **Modularity**, the ability of writing code in separate componentes that work together and can be reused for other programs.

This is help for data engineer as they can reuse code of blocks on larger projects, which helps them by not coding from zero. Also, outdated code needs to be **Refactored**, in other words, reestructure the code while maintaining its original functionality

```python
'''
Binary Search is a searching algorithm that operates on a sorted array, dividing it 
by half to find the target value.

An algorithm is better known as a set of instructions for solving a problem
'''

def binary_search(array, target):
    '''
    Perform binary search to find index target

    Parameters:
        Array (List): Ordered list where it'll search the target
        Target (int): Value to be find

    Returns:
        Mid (int): Index where target has been found
    '''
    low = 0
    high = len(array) - 1

    while low <= high:
        mid = low + (high - low) // 2

        # If target match, return index
        if array[mid] == target:
            return mid
        # If target is greater, ignore left half
        elif mid < target:
            low = mid + 1
        # If target is lower, ignore right half
        else:
            high = mid - 1

    # The target is not found
    return -1
```

```python
array = [2, 3, 4, 10, 40]
target = 10

binary_search(array, target)
```

```python
'''
When comparing text, comparators check the first letter of each string. With A being least 
and Z being the greates 
'''

print("Yellow" > "Cyan" and "Brown" > "Magenta")
```
